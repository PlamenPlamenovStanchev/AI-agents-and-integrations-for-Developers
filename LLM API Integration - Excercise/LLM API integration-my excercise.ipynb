```python
from pydantic import Field
from typing import Literal

class CategorizationResult(BaseModel):
  category: str
 # category: Literal["Invoice", "Contract", "NDA", "Others", "Power of attorney"]
  comment: str = Field(description="Why do yu think this is the most appropriate category")
  confident_level: int = Filed(ge=0, le=100, descripton="The confidence level is a percentage - how sure are you that this is the most precise category for the document")
```

```python
CategorizationResult.model_json_schema()
```
